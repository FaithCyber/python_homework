```python
# task 7
import pandas as pd
import glob
import os
import sys
import subprocess

# 1. Installation and Import
try:
    from thefuzz import process
except ImportError:
    print("Installing thefuzz...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "thefuzz"])
    from thefuzz import process

# 2. Load and Concatenate Data
path = '/kaggle/input/code-the-dream-assignment-6'
all_files = glob.glob(os.path.join(path, "*.csv"))

if not all_files:
    print("No CSV files found. Please check the dataset path.")
else:
    df_list = [pd.read_csv(file) for file in all_files]
    df_all = pd.concat(df_list, axis=0, ignore_index=True)
    print(f"Total rows loaded: {len(df_all)}")

    # 3. Fix Spelling Errors (Fuzzy Matching)
    def fuzzy_cleanup(df, column_name):
        """Identifies common versions of strings and snaps outliers to them."""
        counts = df.value_counts(column_name)
        # We assume values appearing more than 2 times are correctly spelled
        good_values = list(counts[counts > 2].index)
        
        df[column_name] = df[column_name].map(
            lambda x: x if x in good_values else process.extractOne(str(x), good_values)[0]
        )
        return df

    print("Cleaning Name and Address spellings...")
    df_all = fuzzy_cleanup(df_all, "Name")
    df_all = fuzzy_cleanup(df_all, "Address")

    # 4. Fix Zip and Phone (Anomaly Resolution)
    def fix_anomaly(group):
        """Finds the most frequent non-null value (mode) in a group."""
        group_na = group.dropna()
        if group_na.empty:
            return group
        mode = group_na.mode()
        if mode.empty:
            return group
        return mode.iloc[0]

    print("Resolving Zip and Phone inconsistencies...")
    # Group by the (now cleaned) Name and Address to find the "correct" details
    df_all["Zip"] = df_all.groupby(["Name", "Address"])["Zip"].transform(fix_anomaly)
    df_all["Phone"] = df_all.groupby(["Name", "Address"])["Phone"].transform(fix_anomaly)

    # 5. Final Deduplication
    df_final = df_all.drop_duplicates()

    print("-" * 30)
    print(f"Final unique records: {len(df_final)}")
    print("-" * 30)

    # 6. Final Inspection
    df_final.info()
```
